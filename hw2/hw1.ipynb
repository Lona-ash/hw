```python
class Chocolate:
    def __init__(self, price, dop=None):
        if price <= 0:
            raise ValueError
        self.price = price
        self.dop = dop

    def show_my_chocolate(self):
        if self.dop != None:
            return f'Шоколадный батончик, добавка: {self.dop}'
        else:
            return 'Обычный шоколадный батончик'

    def __str__(self):
        return f'Шоколадный батончик, {self.price}'


choco1 = Chocolate(10, "орехи")
print(choco1)
print(choco1.show_my_chocolate())

print()

choco2 = Chocolate(65.50)
print(choco2)
print(choco2.show_my_chocolate())

print()

choco3 = Chocolate(120.05, "карамель")
print(choco3)
print(choco3.show_my_chocolate())

print()

try:
    choco4 = Chocolate(-10, "кокос")
except ValueError:
    print('Ошибка: Цена меньше нуля :( ')
```
