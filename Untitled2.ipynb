Выведите самое популярное мужское и самое популярное женское имя на корабле


```python
male_df = df[df['Sex'] == 'male'].copy()
male_df['FirstName'] = male_df['Name'].str.split(',').str[1].str.split('.').str[1].str.strip().str.split().str[0]

male_df['FirstName'].value_counts().head(1)
```

```python
female_df = df[df['Sex'] == 'female'].copy()

#со скобками те кто жены
with_brackets  = female_df[female_df['Name'].str.contains(r'\(')].copy()
with_brackets ['FirstName'] = with_brackets ['Name'].str.split('(').str[1].str.split(')').str[0].str.strip().str.split().str[0]

# без скобок те кто миссис
without_brackets = female_df.drop(with_brackets.index).copy()
without_brackets['FirstName'] = without_brackets['Name'].str.split(',').str[1].str.split('.').str[1].str.strip().str.split().str[0]

# Объединяем
female_all = pd.concat([with_brackets , without_brackets])
female_all['FirstName'].value_counts().head(1)
```

Выведите самое популярное мужское и самое популярное женское имя на корабле в каждом классе - 2 балла


---



```python
all_df = pd.concat([male_df, female_all])

result = all_df.groupby(['Pclass', 'Sex'])['FirstName'].agg(lambda x: x.value_counts().idxmax())
result
```
