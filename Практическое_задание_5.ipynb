hint: используйте методы: bn.nanmean(), bn.nanstd()

```python
mean_age = bn.nanmean(df['Age'])
std_age = bn.nanstd(df['Age'])


print(f"Средний возраст")
print(mean_age)

print(f"   ")

print(f"Стандартное отклонение")
print(std_age)
```

# Оптимизация типов pandas


Считайте датасет из файла Housing.csv (это данные о ценах домов) с помощью pandas

```python
import pandas as pd
import numpy as np

df = pd.read_csv('Housing.csv')
print(df.head(5))
```

# Для каждого столбца определите оптимальный с точки зрения потребления памяти тип данных - напишите свои выводы в комментариях

```python
print(df.shape)
print(f'  ')

print(df.dtypes)
```

```python
print("Потребление памяти")
print(df.memory_usage(deep=True).sum() / 1024**2, "MB")

```

```python
df = df.convert_dtypes()
print(df.convert_dtypes())
```

```python
yes_no_cols = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea']
for col in yes_no_cols:
    df[col] = df[col].map({'yes': True, 'no': False})

df['furnishingstatus'] = df['furnishingstatus'].astype('category')

df['price'] = pd.to_numeric(df['price'], downcast='unsigned')
df['area'] = pd.to_numeric(df['area'], downcast='unsigned')
df['bedrooms'] = pd.to_numeric(df['bedrooms'], downcast='unsigned')
df['bathrooms'] = pd.to_numeric(df['bathrooms'], downcast='unsigned')
df['stories'] = pd.to_numeric(df['stories'], downcast='unsigned')
df['parking'] = pd.to_numeric(df['parking'], downcast='unsigned')

memory_after = df.memory_usage(deep=True).sum() / 1024**2

print(f"Потребление памяти после: {memory_after:.2f} МБ")
print(f"\nТипы данных после оптимизации:\n{df.dtypes}")
```

## Итог

До: 0.22 МБ

После: 0.01 МБ

До оптимизации DataFrame занимал 0.22 МБ при 545 строках и 13 столбцах. Все числовые столбцы имели тип int64 (8 байт на значение), а строковые столбцы — тип object (около 50 байт на значение).

После оптимизации память сократилась до 0.01МБ. Это означает уменьшение в 20 раз.

Получается, что надо маленькие числа заменить на маленькие типы, бинарные значения -на boolean, повторяющиеся строки( у нас в последнем столбце было всего 3 различных значения) -на категории
